```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("titanic.csv")

print("Dataset loaded successfully!")

print(df.head())
```

```python
# Display number of rows and columns
print("Dataset Shape:", df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns)

# Display dataset information
print("\nDataset Information:")
df.info()

# Display statistical summary
print("\nStatistical Summary:")
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Remove deck column
df.drop(columns=["deck"], inplace=True, errors="ignore")

# Fill numerical missing values with median
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values with mode
for col in df.select_dtypes(include=["str", "object", "category"]).columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values handled successfully!")

print(df.isnull().sum())
```

```python
print("Duplicates before cleaning:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicates after cleaning:", df.duplicated().sum())
```

```python
categorical_columns = [
    "sex",
    "embarked",
    "class",
    "who",
    "embark_town",
    "alive"
]

for col in categorical_columns:
    if col in df.columns:
        df[col] = df[col].astype("category")

# Convert passenger class into categorical type
if "pclass" in df.columns:
    df["pclass"] = df["pclass"].astype("category")

print("Categorical columns converted successfully!")

print(df.dtypes)
```
